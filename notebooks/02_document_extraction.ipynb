Read the sinhala document

```python
from pathlib import Path

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

document_path = project_root / "data" / "raw" / "ella.txt"

with open(document_path, "r", encoding="utf-8-sig") as file:
    document_text = file.read()

print("Document loaded successfully!")
print("Total characters:", len(document_text))
print("\nDocument preview:\n")
print(document_text[:1000])
```

Identify document sections

```python
import re

section_pattern = r"(?m)^##\s+\d+\.\s+(.+)$"

sections = re.findall(section_pattern, document_text)

print("Total sections:", len(sections))

for index, section in enumerate(sections, 1):
    print(f"{index}. {section}")
```

Create the relationship dataset

```python
relationships = [
    {"source": "D001", "relation": "LOCATED_IN", "target": "DI001"},
    {"source": "DI001", "relation": "PART_OF", "target": "P001"},

    {"source": "D001", "relation": "HAS_ATTRACTION", "target": "A001"},
    {"source": "D001", "relation": "HAS_ATTRACTION", "target": "A002"},
    {"source": "D001", "relation": "HAS_ATTRACTION", "target": "A003"},
    {"source": "D001", "relation": "HAS_ATTRACTION", "target": "A004"},
    {"source": "D001", "relation": "HAS_ATTRACTION", "target": "A005"},
    {"source": "D001", "relation": "HAS_ATTRACTION", "target": "A006"},
    {"source": "D001", "relation": "HAS_ATTRACTION", "target": "A007"},
    {"source": "D001", "relation": "HAS_ATTRACTION", "target": "A008"},

    {"source": "D001", "relation": "HAS_ACTIVITY", "target": "AC001"},
    {"source": "D001", "relation": "HAS_ACTIVITY", "target": "AC002"},
    {"source": "D001", "relation": "HAS_ACTIVITY", "target": "AC003"}
]

relationships_df = pd.DataFrame(relationships)

display(relationships_df)

print("Total relationships:", len(relationships_df))
```

Validate the extracted data

```python
entity_ids = set(entities_df["id"])

invalid_relationships = relationships_df[
    ~relationships_df["source"].isin(entity_ids)
    | ~relationships_df["target"].isin(entity_ids)
]

print("Invalid relationships:", len(invalid_relationships))

if invalid_relationships.empty:
    print("Validation passed!")
else:
    display(invalid_relationships)
```

Save the extracted data

```python
from pathlib import Path

output_dir = project_root / "data" / "extracted"
output_dir.mkdir(parents=True, exist_ok=True)

entities_df.to_csv(
    output_dir / "entities.csv",
    index=False,
    encoding="utf-8-sig"
)

relationships_df.to_csv(
    output_dir / "relationships.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Entity and relationship datasets saved successfully!")
```
